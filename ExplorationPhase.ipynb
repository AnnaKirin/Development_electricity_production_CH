```python
df = pd.read_csv(
    "production_data_from_1900.csv",
    delimiter=",",
)

Power_by_technology = df.groupby(["BeginningOfOperation", "Canton", "MainCategory"], as_index=False).agg(
    TotalPower=("TotalPower", "sum"))
print(Power_by_technology)
```

```python
chart = alt.Chart(Power_by_technology).mark_area().encode(
    x=alt.X("BeginningOfOperation:Q", title="Year"),
    y=alt.Y("TotalPower:Q", title="Power added (kW)"),
    color=alt.Color("MainCategory:N", title="Technology")
)

chart.save("power_by_technology.html")
```

```python
CumulativePower = Power_by_technology.sort_values(["BeginningOfOperation"])
CumulativePower["InstalledPower"] = (CumulativePower["TotalPower"].cumsum())
#Nawias ( ) wywołuje funkcję. Nawias [ ] wybiera kolumnę

CumulativePower
```

```python
#--Line illustrating the running total of Power

chart = alt.Chart(CumulativePower).mark_line().encode(
    x="BeginningOfOperation:Q",
    y="InstalledPower:Q", )

chart.save("power_by_technology.html")
```

```python
#--Bars + line

base = alt.Chart(CumulativePower).encode(
    x=alt.X("BeginningOfOperation:Q", title="Year", axis=alt.Axis(format="d"))
)
bars = base.mark_bar().encode(
    y=alt.Y("TotalPower:Q", title="Power added (kW)")
)
line = base.mark_line(color='blue').encode(
    y=alt.Y("InstalledPower:Q", title="Power installed (kW)")
)
chart = (bars+line).resolve_scale(y="independent")
chart.save("power_by_technology.html")

```

```python
#--Area + Line - incomprehensible

base = alt.Chart(CumulativePower).encode(
    x=alt.X("BeginningOfOperation:Q", title="Year", axis=alt.Axis(format="d"))
)
area = alt.Chart(Power_by_technology).mark_area().encode(
    x=alt.X("BeginningOfOperation:Q", title="Year"),
    y=alt.Y("TotalPower:Q", title="Power added (kW)"),
    color=alt.Color("MainCategory:N", title="Technology")
)
line = base.mark_line(color='blue').encode(
    y=alt.Y("InstalledPower:Q", title="Power installed (kW)")
)
chart = (area+line).resolve_scale(y="independent")
chart.save("power_by_technology.html")
```
